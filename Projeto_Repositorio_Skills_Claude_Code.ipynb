## 4. 🛠️ Skill Builder — Criando uma Skill passo a passo

O **Skill Builder** é uma função Python que recebe os parâmetros de uma Skill e gera automaticamente sua estrutura padrão:

```
Nome da Skill → Descrição → Categoria → Objetivo → Quando utilizar → Entradas → Processo → Saída esperada
                                        ↓
                          skills/<categoria>/<nome>/SKILL.md
```


````python
import re

def slugify(nome: str) -> str:
    """Converte o nome da Skill para kebab-case (ex.: 'Pandas Analysis' -> 'pandas-analysis')."""
    s = nome.strip().lower()
    s = re.sub(r"[^a-z0-9]+", "-", s)
    s = re.sub(r"-+", "-", s).strip("-")
    return s


def build_skill_md(nome, descricao, objetivo, quando_usar, entradas, processo, saida_esperada, regras=None):
    """Monta o conteúdo textual do SKILL.md a partir dos parâmetros do Skill Builder."""
    slug = slugify(nome)
    regras = regras or [
        "Preservar os dados/contexto originais.",
        "Explicar transformações e decisões tomadas.",
        "Informar problemas ou limitações encontradas.",
    ]

    processo_fmt = "\n".join(f"{i}. {passo}" for i, passo in enumerate(processo, 1))
    regras_fmt = "\n".join(f"- {r}" for r in regras)
    entradas_fmt = "\n".join(f"- {e}" for e in entradas)

    # Construído por linhas (em vez de textwrap.dedent) para não depender da
    # indentação do bloco onde a função é chamada.
    linhas = [
        "---",
        f"name: {slug}",
        f"description: {descricao}",
        "---",
        "",
        f"# {nome}",
        "",
        "## Objetivo",
        objetivo,
        "",
        "## Quando usar",
        quando_usar,
        "",
        "## Entradas",
        entradas_fmt,
        "",
        "## Procedimento",
        processo_fmt,
        "",
        "## Saída esperada",
        saida_esperada,
        "",
        "## Regras",
        regras_fmt,
        "",
    ]
    content = "\n".join(linhas)
    return slug, content


def build_skill(nome, descricao, categoria, objetivo, quando_usar, entradas, processo,
                 saida_esperada, regras=None, root=REPO_ROOT, overwrite=False):
    """Skill Builder: gera skills/<categoria>/<nome-slug>/SKILL.md e retorna o caminho criado."""
    if categoria not in CATEGORIAS_VALIDAS:
        raise ValueError(f"Categoria inválida: {categoria!r}. Use uma de {CATEGORIAS_VALIDAS}")

    slug, content = build_skill_md(nome, descricao, objetivo, quando_usar, entradas, processo,
                                    saida_esperada, regras)

    skill_dir = os.path.join(root, "skills", categoria, slug)
    os.makedirs(skill_dir, exist_ok=True)
    skill_path = os.path.join(skill_dir, "SKILL.md")

    if os.path.exists(skill_path) and not overwrite:
        raise FileExistsError(f"{skill_path} já existe. Use overwrite=True para sobrescrever.")

    with open(skill_path, "w", encoding="utf-8") as f:
        f.write(content)

    print(f"✅ Skill criada em: {skill_path}")
    return skill_path
````

## 5. ✅ Validador Automático

Verifica a qualidade estrutural de cada Skill em `skills/` antes do commit:

- ✓ `SKILL.md` existente.
- ✓ Frontmatter/YAML válido.
- ✓ Nome e descrição preenchidos.
- ✓ Estrutura de diretórios válida.
- ✓ Documentação mínima presente.
- ✓ Arquivos auxiliares acessíveis.
- ✓ Links e referências verificáveis (checagem básica).


````python
import glob

MIN_BODY_LENGTH = 80  # caracteres mínimos no corpo do SKILL.md (após o frontmatter)


def parse_frontmatter(text):
    """Faz um parse simples de frontmatter YAML (--- ... ---) sem dependências externas."""
    m = re.match(r"^---\s*\n(.*?)\n---\s*\n(.*)$", text, re.DOTALL)
    if not m:
        return None, text
    raw_yaml, body = m.group(1), m.group(2)
    data = {}
    for line in raw_yaml.splitlines():
        line = line.strip()
        if not line or line.startswith("#") or ":" not in line:
            continue
        key, _, value = line.partition(":")
        data[key.strip()] = value.strip()
    return data, body


def validate_skill(skill_dir, root=REPO_ROOT):
    """Valida uma única Skill (pasta). Retorna (ok: bool, erros: list[str], avisos: list[str])."""
    erros, avisos = [], []
    rel = os.path.relpath(skill_dir, root)

    skill_md = os.path.join(skill_dir, "SKILL.md")
    if not os.path.isfile(skill_md):
        return False, [f"[{rel}] SKILL.md não encontrado."], []

    with open(skill_md, encoding="utf-8") as f:
        text = f.read()

    frontmatter, body = parse_frontmatter(text)
    if frontmatter is None:
        erros.append(f"[{rel}] Frontmatter YAML ausente ou malformado (esperado bloco --- ... ---).")
        frontmatter = {}

    nome = frontmatter.get("name", "").strip()
    descricao = frontmatter.get("description", "").strip()

    if not nome:
        erros.append(f"[{rel}] Campo 'name' vazio ou ausente no frontmatter.")
    elif not re.match(r"^[a-z0-9]+(-[a-z0-9]+)*$", nome):
        avisos.append(f"[{rel}] 'name' ({nome!r}) não está em kebab-case.")

    if not descricao:
        erros.append(f"[{rel}] Campo 'description' vazio ou ausente no frontmatter.")
    elif len(descricao) < 10:
        avisos.append(f"[{rel}] 'description' muito curta — detalhe o que a Skill faz e quando usar.")

    # categoria válida a partir do caminho skills/<categoria>/<slug>/
    partes = rel.split(os.sep)
    if len(partes) >= 2 and partes[0] == "skills":
        categoria = partes[1]
        if categoria not in CATEGORIAS_VALIDAS:
            avisos.append(f"[{rel}] Categoria '{categoria}' fora da lista padrão {CATEGORIAS_VALIDAS}.")

    if len(body.strip()) < MIN_BODY_LENGTH:
        erros.append(f"[{rel}] Documentação insuficiente no corpo do SKILL.md (mínimo {MIN_BODY_LENGTH} caracteres).")

    # arquivos auxiliares referenciados (ex.: scripts/, references/, assets/) devem existir
    for m in re.finditer(r"`([\w./-]+\.[\w]+)`", body):
        ref = m.group(1)
        if ref.startswith(("http://", "https://")):
            continue
        candidate = os.path.join(skill_dir, ref)
        if "/" in ref and not os.path.exists(candidate) and not os.path.exists(os.path.join(root, ref)):
            avisos.append(f"[{rel}] Referência de arquivo '{ref}' mencionada no SKILL.md não foi encontrada.")

    ok = len(erros) == 0
    return ok, erros, avisos


def validate_all_skills(root=REPO_ROOT):
    """Roda validate_skill em todas as Skills encontradas em skills/*/*/SKILL.md."""
    pattern = os.path.join(root, "skills", "*", "*")
    skill_dirs = sorted(d for d in glob.glob(pattern) if os.path.isdir(d))

    resultados = []
    for skill_dir in skill_dirs:
        ok, erros, avisos = validate_skill(skill_dir, root=root)
        resultados.append((skill_dir, ok, erros, avisos))
    return resultados
````

````python
resultados = validate_all_skills()

if not resultados:
    print("⚠️  Nenhuma Skill encontrada em skills/*/*/SKILL.md.")
    print("   Rode a seção 4 (Skill Builder) primeiro para criar uma Skill de exemplo.")
else:
    total_ok = sum(1 for _, ok, _, _ in resultados if ok)
    print(f"Validando {len(resultados)} Skill(s)...\n")

    for skill_dir, ok, erros, avisos in resultados:
        status = "✅ OK" if ok else "❌ FALHOU"
        print(f"{status}  {os.path.relpath(skill_dir, REPO_ROOT)}")
        for e in erros:
            print(f"    ERRO: {e}")
        for a in avisos:
            print(f"    AVISO: {a}")

    print(f"\nResumo: {total_ok}/{len(resultados)} Skill(s) passaram na validação.")
    if total_ok < len(resultados):
        print("Corrija os erros acima antes de prosseguir para o commit / publicação.")
````

## 6. 🗂️ Catálogo Automático

Lê os metadados (frontmatter) de todas as Skills em `skills/`, monta uma tabela por categoria e atualiza automaticamente a seção de catálogo do `README.md`, entre os marcadores:

```
<!-- CATALOG:START -->
...
<!-- CATALOG:END -->
```


````python
README_PATH = os.path.join(REPO_ROOT, "README.md")


def collect_skills(root=REPO_ROOT):
    """Varre skills/<categoria>/<slug>/SKILL.md e retorna metadados agrupados por categoria."""
    por_categoria = {cat: [] for cat in CATEGORIAS_VALIDAS}

    for skill_md in sorted(glob.glob(os.path.join(root, "skills", "*", "*", "SKILL.md"))):
        categoria = os.path.basename(os.path.dirname(os.path.dirname(skill_md)))
        with open(skill_md, encoding="utf-8") as f:
            text = f.read()
        meta, _ = parse_frontmatter(text)
        meta = meta or {}
        nome = meta.get("name", os.path.basename(os.path.dirname(skill_md)))
        descricao = meta.get("description", "")
        rel_path = os.path.relpath(skill_md, root).replace(os.sep, "/")
        por_categoria.setdefault(categoria, []).append((nome, descricao, rel_path))

    return por_categoria


def render_catalog_markdown(por_categoria):
    linhas = []
    total = sum(len(v) for v in por_categoria.values())
    if total == 0:
        return "_Nenhuma Skill publicada ainda. Este catálogo será preenchido conforme Skills forem adicionadas em `skills/`._"

    for categoria in CATEGORIAS_VALIDAS:
        skills = por_categoria.get(categoria, [])
        if not skills:
            continue
        linhas.append(f"### {categoria.capitalize()}\n")
        linhas.append("| Skill | Descrição |")
        linhas.append("|---|---|")
        for nome, descricao, rel_path in skills:
            linhas.append(f"| [`{nome}`]({rel_path}) | {descricao} |")
        linhas.append("")
    return "\n".join(linhas).strip()


DEFAULT_README = """# claude-code-skills-repository

## Catálogo de Skills

<!-- CATALOG:START -->
_Nenhuma Skill publicada ainda._
<!-- CATALOG:END -->
"""


def ensure_readme_with_markers(readme_path=README_PATH):
    """Garante que exista um README.md com os marcadores de catálogo (útil ao rodar este
    notebook de forma independente, sem o repositório clonado)."""
    os.makedirs(os.path.dirname(readme_path), exist_ok=True)
    if not os.path.isfile(readme_path):
        with open(readme_path, "w", encoding="utf-8") as f:
            f.write(DEFAULT_README)
        print(f"README.md não encontrado — criado um novo em {readme_path}")


def generate_catalog(root=REPO_ROOT, readme_path=README_PATH, write=True):
    """Gera o catálogo e substitui o bloco entre <!-- CATALOG:START/END --> no README.md."""
    por_categoria = collect_skills(root)
    catalog_md = render_catalog_markdown(por_categoria)

    ensure_readme_with_markers(readme_path)
    with open(readme_path, encoding="utf-8") as f:
        readme = f.read()

    start_marker, end_marker = "<!-- CATALOG:START -->", "<!-- CATALOG:END -->"
    if start_marker not in readme or end_marker not in readme:
        raise ValueError("Marcadores de catálogo não encontrados no README.md")

    before = readme.split(start_marker)[0]
    after = readme.split(end_marker)[1]
    novo_readme = f"{before}{start_marker}\n{catalog_md}\n{end_marker}{after}"

    if write:
        with open(readme_path, "w", encoding="utf-8") as f:
            f.write(novo_readme)

    return catalog_md, novo_readme
````

````python
catalog_md, novo_readme = generate_catalog(write=True)

print("Catálogo gerado a partir das Skills encontradas em skills/*/*/SKILL.md:\n")
print(catalog_md)
print("\n✅ README.md atualizado entre os marcadores <!-- CATALOG:START --> e <!-- CATALOG:END -->.")
````
